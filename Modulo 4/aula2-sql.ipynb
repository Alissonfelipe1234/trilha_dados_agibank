```python
import sqlite3
import pandas as pd

DB_PATH = 'escola_v2.db'  # Caminho para o banco de dados SQLite
```

```python
def executar_query(sql_query):
    try:
        conn = sqlite3.connect(DB_PATH)
        df_resultado = pd.read_sql_query(sql_query, conn)
        conn.close()
        return df_resultado
    except Exception as e:
        print("Erro:", e)
```
