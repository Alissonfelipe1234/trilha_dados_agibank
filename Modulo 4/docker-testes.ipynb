```python
import pandas as pd
from sqlalchemy import create_engine, text


# 1. Configurando a conexão com o Postgres no Docker
connection_string = 'postgresql://postgres:senha123@localhost:5432/postgres'
engine = create_engine(connection_string)
```

```python
# 2. Criando um DataFrame simples
df_teste = pd.DataFrame({
    'id': [1, 2, 3, 5],
    'nome': ['Ana', 'Gustavo', 'Henrique', 'Patrik'],
    'cargo': ['CEO', 'Analista', 'Gerente de Projetos', 'Chefe']
})
df_teste
```

```python
cursor = engine.connect()
cursor.execute(text('''TRUNCATE TABLE funcionarios''' ))
cursor.close()
```

```python
# 3. Escrevendo no Banco
df_teste.to_sql('funcionarios', 
                engine, 
                if_exists='replace', 
                index=False)
```

```python
# 4. Select dados
df_lido = pd.read_sql(
    "SELECT * FROM funcionarios", 
    engine)
df_lido
```
