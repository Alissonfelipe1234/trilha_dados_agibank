```python
from transformers import pipeline
import pandas as pd
import numpy as np
```

```python
model_leve = "./modelo_sentimento_offline"

analisador_leve = pipeline(
    "sentiment-analysis", 
    model=model_leve, 
    tokenizer=model_leve
)
```

```python
frase = 'pode melhorar esse produto, nada de mais'
analisador_leve(frase)
```

```python
frase = "O atendimento foi péssimo, demoraram muito!"
resultado = analisador_leve(frase)

print(f"Texto: {frase}")
print(f"Resultado: {resultado}")
```

```python
def analisar_completo(text):
    retorno = analisador_leve(text)
    return [
        retorno[0]['label'], retorno[0]['score']
    ]

reviews[['sentimento', 'score']] = reviews['texto']\
    .apply(lambda x: pd.Series(\
        analisar_completo(x)
    ))
```

```python
# Tentativa Ana
reviews = pd.read_csv('review.csv')
reviews
```

```python
reviews["score"] = np.nan
reviews["sentimento"] = np.nan
```

```python
for index, row in reviews.iterrows():
    try:
        scores = analisador_leve(row['texto'])
        reviews.loc[index, 'sentimento'] = scores[0]['label']
        reviews.loc[index, 'score'] = scores[0]['score']
    except:
        print("ERROR")
        raise Exception
```

```python
def analisar_sentimento(texto, model=analisador_leve):
    resultado = model(texto)
    return resultado[0]['label']
```

```python
reviews['sentimento'] = reviews['texto'].apply(analisar_sentimento)
```
