```python
import requests
import json

url = "http://localhost:11434/api/generate"
modelo = "deepseek-r1:8b" # Tem que ser o mesmo nome que baixamos
```

```python

def perguntar_ao_ollama(prompt):
    payload = {
        "model": modelo,
        "prompt": prompt,
        "stream": False,
        "temperature": 0.5 # Criatividade
    }
    
    print("Processando...")
    response = requests.post(url, json=payload)
    
    if response.status_code == 200:
        return response.json()['response']
    else:
        return f"Erro: {response.text}"

```

```python
pergunta = input("diga o que quer perguntar:")
resposta = perguntar_ao_ollama(pergunta)

print("-" * 40)
print(resposta)
print("-" * 40)
```
