```python
# Script Gerador de CSV Gigante (Fake)
from pyspark.sql import SparkSession
from pyspark.sql.functions import expr

spark = SparkSession.builder.appName("Gerador")\
.master("local[*]").getOrCreate()

df_fake = spark.range(0, 80000000).select(
    expr("id"),
    expr("rand() * 1000 as valor"),
    expr("case when rand() > 0.5 then 'Aprovado' else 'Reprovado' end as status"),
    expr("uuid() as transacao_hash") # Texto longo
)

print("Salvando CSV gigante...")
df_fake.write.csv("dados/arquivo_gigante.csv", header=True, mode="overwrite")
print("pronto!")
```
