```python
import os
import shutil
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma


NOME_DO_ARQUIVO = "beneficios.pdf" 
PASTA_DOCUMENTOS = "./"
PASTA_BANCO = "./banco_vetorial"

caminho_pdf = os.path.join(PASTA_DOCUMENTOS, NOME_DO_ARQUIVO)
```

```python
print("Apagando banco de dados antigo para evitar conflitos...")
if os.path.exists(PASTA_BANCO):
    shutil.rmtree(PASTA_BANCO)
    print("   -> Banco antigo apagado.")
else:
    print("   -> Nenhum banco encontrado (criando do zero).")


if not os.path.exists(caminho_pdf):
    print(f"ERRO: O arquivo não está em '{caminho_pdf}'")
    exit()
```

```python

loader = PyPDFLoader(caminho_pdf)
documentos = loader.load()

print(f"Sucesso! {len(documentos)} páginas carregadas.")


print("\n--- [INÍCIO DO TEXTO LIDO] ---")
print(documentos[0].page_content[:200]) 
print("--- [FIM DO TEXTO LIDO] ---\n")


text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    separators=["\n\n", "\n", " ", ""] 
)

chunks = text_splitter.split_documents(documentos)
print(f"Criados {len(chunks)} pedaços (chunks).")
```

```python
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# Cria e persiste automaticamente
vector_store = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    persist_directory=PASTA_BANCO
)

print("SUCESSO! O banco foi recriado.")
```
